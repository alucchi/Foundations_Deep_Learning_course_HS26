```python
import torch
import matplotlib.pyplot as plt
import numpy as np
from transformers import GPT2Tokenizer, GPT2Model

input_text = "Like a fish out of water"

def show_token_embedding_norms_and_vectors_from_text(text):
    # Load GPT-2 tokenizer and model
    tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
    model = GPT2Model.from_pretrained("gpt2")

    # Tokenize the input text
    inputs = tokenizer(text, return_tensors="pt", add_special_tokens=False)
    input_ids = inputs["input_ids"][0]

    # Get input embeddings for the tokens
    with torch.no_grad():
        embeddings = model.wte(input_ids)  # shape: (T, d)
        norms = torch.norm(embeddings, dim=1).tolist()
        embeddings_list = embeddings.tolist()

    # Display result
    print(f"Input text: {repr(text)}\n")
    print(f"{'Token':<20} {'Token ID':<10} {'L2 Norm':<10} {'Embedding Vector'}")
    print("-" * 90)
    for token, token_id, norm, vector in zip(tokenizer.convert_ids_to_tokens(input_ids), input_ids, norms, embeddings_list):
        vector_str = "[" + ", ".join(f"{v:.3f}" for v in vector[:8]) + ("..." if len(vector) > 8 else "") + "]"
        print(f"{token:<20} {token_id:<10} {norm:<10.4f} {vector_str}")

show_token_embedding_norms_and_vectors_from_text(input_text)
```
